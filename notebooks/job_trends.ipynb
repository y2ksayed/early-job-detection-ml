# Job Trends Exploration

Quick EDA over the sample postings: category mix, skill frequency, and posting velocity.
Run the scraper first to get real data, or use `data/sample_jobs.json` as-is.

```python
import json
from collections import Counter

with open("../data/sample_jobs.json") as f:
    jobs = json.load(f)

print(f"Loaded {len(jobs)} postings")
```

```python
# Postings by category
from collections import Counter

by_category = Counter(j["category"] for j in jobs)
for cat, n in by_category.most_common():
    print(f"{cat:25s} {n}")
```

```python
# Most common skills
skills = Counter(s.lower() for j in jobs for s in j.get("skills", []))
print("Top skills:")
for skill, n in skills.most_common(10):
    print(f"  {skill:15s} {n}")
```

```python
# Emerging vs repost, urgent vs normal
emerging = sum(1 for j in jobs if j.get("is_emerging"))
urgent = sum(1 for j in jobs if j.get("is_urgent"))
print(f"Emerging: {emerging}/{len(jobs)}")
print(f"Urgent:   {urgent}/{len(jobs)}")

# Repost check: same title+company appearing more than once
seen = Counter((j["title"], j["company"]) for j in jobs)
reposts = sum(1 for c in seen.values() if c > 1)
print(f"Reposted title+company pairs: {reposts}")
```
